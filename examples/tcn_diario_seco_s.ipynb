## Importações

```python
%matplotlib inline

import warnings

import matplotlib.pyplot as plt
import pandas as pd

from darts import TimeSeries, concatenate
from darts.dataprocessing.transformers import Scaler
from darts.models import TCNModel
from darts.utils.callbacks import TFMProgressBar
from darts.utils.timeseries_generation import datetime_attribute_timeseries


warnings.filterwarnings("ignore")

def generate_torch_kwargs():
    # run torch models on CPU, and disable progress bars for all model stages except training.
    return {
        "pl_trainer_kwargs": {
            "accelerator": "cpu",
            "callbacks": [TFMProgressBar(enable_train_bar_only=True)],
        }
    }
```

## Carregamento de dados

```python
dataframe_seco_s_diario = pd.read_csv("INTERCAMBIO_SUDESTE-CENTRO-OESTE_SUL_DIARIO_2024-2025.csv", sep=";", encoding="utf-8")

dataframe_seco_s_diario["din_instante"] = pd.to_datetime(dataframe_seco_s_diario["din_instante"])

dataframe_seco_s_diario.info()
```

## Carrega dataset

```python
# Define dataset
range_train_val = pd.date_range(start="2024-01-01 00:00:00", end="2025-09-30 23:59:00", freq="D")
range_compare = pd.date_range(start="2025-10-01", end="2025-10-31", freq="D")

dataframe_seco_s_diario_set2025 = dataframe_seco_s_diario[dataframe_seco_s_diario['din_instante'].isin(range_train_val)]

dataframe_seco_s_diario_compare = dataframe_seco_s_diario[dataframe_seco_s_diario['din_instante'].isin(range_compare)]
```

```python
# Criar TimeSeries
ts = TimeSeries.from_dataframe(dataframe_seco_s_diario_set2025, time_col="din_instante", value_cols=["val_intercambiomwmed"], freq="D")

# Create training and validation sets:
train, val = ts.split_after(pd.Timestamp("2025-04-30 23:59:00"))

scaler = Scaler()

train_scaled = scaler.fit_transform(train)
val_scaled = scaler.transform(val)
ts_scaled = scaler.transform(ts)
```

```python
# Adicionar a covariável dia do mês (scaling not required as one-hot-encoded)
day_series = datetime_attribute_timeseries(
    scaler.transform(TimeSeries.from_dataframe(dataframe_seco_s_diario, time_col="din_instante", value_cols=["val_intercambiomwmed"], freq="D")),
    attribute="day",
    one_hot=True
)
```

```python
plt.figure(figsize=(10, 3))
train_scaled.plot(label="Treino")
val_scaled.plot(label="Validação")
plt.title('Conjunto de Treino e Validação (Escalonado) - Intercâmbio Sudeste/Centro-Oeste -> Sul')
plt.xlabel('Data')
plt.legend()
```

## Construir modelo

```python
model_name = "TCN_Intercambio_SECO_S_Diario_2024_2025"
my_model = TCNModel(
    input_chunk_length=56, # 8 semanas
    output_chunk_length=28,  # 4 semanas
    n_epochs=30,
    dropout=0.2,
    dilation_base=2,
    weight_norm=True,
    kernel_size=7,
    num_filters=28,
    nr_epochs_val_period=1,
    random_state=0,
    save_checkpoints=True,
    model_name=model_name,
    force_reset=True,
    **generate_torch_kwargs(),
)
```

## Treinar Modelo

```python
my_model.fit(
    series=train_scaled,
    past_covariates=day_series,
    val_series=val_scaled,
    val_past_covariates=day_series,
)
```

## Carregar Modelo

```python
my_model = TCNModel.load_from_checkpoint(model_name=model_name, best=True)
```

## Backtest

```python
backtest = my_model.historical_forecasts(
    series=ts_scaled,
    past_covariates=day_series,
    start=val_scaled.start_time(),
    forecast_horizon=28, # 4 semanas
    stride=28, # 4 semanas
    last_points_only=False,
    retrain=False,
    verbose=True,
)
backtest = concatenate(backtest)
```

```python
plt.figure(figsize=(10, 6))
val_scaled.plot(label="Conjunto de Validação")
backtest.plot(label="Backtest (H=28)")
plt.title('Backtest Com Horizonte de 4 Semanas - Intercâmbio Sudeste/Centro-Oeste -> Sul')
plt.xlabel('Data')
plt.legend()
```

```python
from darts.metrics import mae, mse

print(f"MAE: {mae(val_scaled,backtest)}")
print(f"MSE: {mse(val_scaled,backtest)}")
```

```python
# dados de outubro
real = TimeSeries.from_dataframe(dataframe_seco_s_diario_compare, time_col="din_instante", value_cols=["val_intercambiomwmed"], freq="D")

# Previsão para o próximo mês (outubro de 2025)
pred_scaled = my_model.predict(
    n=28,
    series=ts,               # A partir de onde prever (fim do treino+validação)
    past_covariates=day_series # Covariáveis disponíveis (precisa conter os valores para o horizonte de previsão)
)
```

```python
predicted = pred_scaled

plt.figure(figsize=(10, 6))
real.plot(label="Verificado")
pred_scaled.plot(label="Previsto (H=28)")
plt.title('Previsão TCN (4 primeiras semanas Out/2025) - Intercâmbio Sudeste/Centro-Oeste -> Sul')
plt.ylabel('Intercâmbio (MWmed)')
plt.xlabel('Data')
plt.legend()
```
