## Importação de dados

```python
dados = pd.read_csv("INTERCAMBIO_NACIONAL_2024-2025.csv", sep=";", encoding="utf-8")
dados
```

## Preparação inicial

```python
# Excluir colunas "id_subsistema_origem" e "id_subsistema_destino"
dataframe = dados.drop(columns=["id_subsistema_origem", "id_subsistema_destino"], errors="ignore")

# Substituir valores de duas colunas SUDESTE -> SUDESTE/CENTRO-OESTE
dataframe['nom_subsistema_origem'] = dataframe['nom_subsistema_origem'].replace({'SUDESTE': 'SUDESTE/CENTRO-OESTE'})
dataframe['nom_subsistema_destino'] = dataframe['nom_subsistema_destino'].replace({'SUDESTE': 'SUDESTE/CENTRO-OESTE'})

# Converter coluna de timestamp para datetime
dataframe['din_instante'] = pd.to_datetime(dataframe['din_instante'])

dataframe.info()

# Verificar frequência e valores ausentes
timedeltas = dataframe['din_instante'].diff().value_counts()
print("\nDistribuição de intervalos entre registros:")
print(timedeltas)
```

## Separação Origem -> Destino

```python
## Dataframes separados com base em "nom_subsistema_origem" e "nom_subsistema_destino"
dataframe_n_ne = dataframe[(dataframe['nom_subsistema_origem'] == 'NORTE') & (dataframe['nom_subsistema_destino'] == 'NORDESTE')].reset_index(drop=True)
dataframe_n_seco = dataframe[(dataframe['nom_subsistema_origem'] == 'NORTE') & (dataframe['nom_subsistema_destino'] == 'SUDESTE/CENTRO-OESTE')].reset_index(drop=True)
dataframe_ne_seco = dataframe[(dataframe['nom_subsistema_origem'] == 'NORDESTE') & (dataframe['nom_subsistema_destino'] == 'SUDESTE/CENTRO-OESTE')].reset_index(drop=True)
dataframe_seco_s = dataframe[(dataframe['nom_subsistema_origem'] == 'SUDESTE/CENTRO-OESTE') & (dataframe['nom_subsistema_destino'] == 'SUL')].reset_index(drop=True)

# Exportar em formato CSV
dataframe_n_ne.to_csv("INTERCAMBIO_NORTE_NORDESTE_2024-2025.csv", sep=";", index=False, encoding="utf-8")
dataframe_n_seco.to_csv("INTERCAMBIO_NORTE_SUDESTE-CENTRO-OESTE_2024-2025.csv", sep=";", index=False, encoding="utf-8")
dataframe_ne_seco.to_csv("INTERCAMBIO_NORDESTE_SUDESTE-CENTRO-OESTE_2024-2025.csv", sep=";", index=False, encoding="utf-8")
dataframe_seco_s.to_csv("INTERCAMBIO_SUDESTE-CENTRO-OESTE_SUL_2024-2025.csv", sep=";", index=False, encoding="utf-8")
```
